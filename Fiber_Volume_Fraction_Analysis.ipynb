# Variation of Fiber Volume Fraction

```python
import pandas as pd
import numpy as np
node_index = pd.read_csv("sample_node_index.csv", sep = ";", index_col = 'node_id')
node_index = node_index[['x', 'y', 'curve_id', 'is_boundary']]
print(node_index.head())
```

```python
df = node_index[node_index['is_boundary'] == True]
print(df['curve_id'].unique())
```

```python
df_bottom = df[df['curve_id'] != 24]
df_top = df[df['curve_id'] != 0]

df_bottom = df_bottom[['x', 'y', 'curve_id']].reset_index()
df_top = df_top[['x', 'y', 'curve_id']].reset_index()

df_bottom.columns = ['bot_node', 'x_bot', 'y_bot', 'bot_curve_id']
df_top.columns = ['top_node', 'x_top', 'y_top', 'top_curve_id']

df_fiber_volume_fraction = pd.concat([df_bottom, df_top], axis = 1)
df_fiber_volume_fraction['local_thickness'] = df_fiber_volume_fraction['y_top'] - df_fiber_volume_fraction['y_bot'] 

df_fiber_volume_fraction['fiber_volume_fraction'] = 0.075 / df_fiber_volume_fraction['local_thickness']
print(df_fiber_volume_fraction['fiber_volume_fraction'].head())
```

```python
stiffness_matrix = 4670
stiffness_fiber = 209053
stiffness_ideal = 127300

df_fiber_volume_fraction['stiffness_corrected'] = (stiffness_fiber * df_fiber_volume_fraction['fiber_volume_fraction'] +
                                                   stiffness_matrix * (1 - df_fiber_volume_fraction['fiber_volume_fraction'])
                                                  )
print(df_fiber_volume_fraction['stiffness_corrected'].max())
```
